# Datos del hackatón por fecha

```python
from pathlib import Path

import pandas as pd

carpeta_datos = Path("Informacion_RETO/Datos")
df = pd.read_csv(carpeta_datos / "datos_hackaton.csv")
config = pd.read_csv(carpeta_datos / "config_hackaton.csv")
```

## Una tabla por fecha

Se interpreta `Week` como semana ISO y `Jour` como día de la semana (1 = lunes). Cada fecha apunta a un DataFrame con las máquinas (`Operacion`) en el orden de `config_hackaton.csv`.

```python
fechas_iso = (
    df["Year"].astype(str)
    + "-W"
    + df["Week"].astype(str).str.zfill(2)
    + "-"
    + df["Jour"].astype(str)
)
datos = df.assign(Fecha=pd.to_datetime(fechas_iso, format="%G-W%V-%u"))
datos = datos.merge(config, on=["Linea", "Operacion"], validate="many_to_one", how="left")
datos = datos.sort_values(["Fecha", "Orden_OP"])

variables = ["TBF", "TAP", "DES", "SAT", "NAP", "NPR"]
tablas_por_fecha = pd.DataFrame(
    [
        (fecha, grupo.set_index("Operacion")[variables])
        for fecha, grupo in datos.groupby("Fecha")
    ],
    columns=["Fecha", "Tabla"],
).set_index("Fecha")
```

## Consultar las fechas disponibles
